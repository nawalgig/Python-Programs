```python
#Product Prices:
product_prices = [3500,2500,6000,1000,2000]

#Increase the price of two products due to tax (1% tax /product).
product_prices[-1] = product_prices[-1]*1.01
product_prices[-2] = product_prices[-2]*1.01

product_prices
```

```python
#Remove the lowest-priced product.
lowest_price = min(product_prices)
product_prices.remove(lowest_price)
product_prices
```

```python
#Calculate the total price of the remaining products.
total_price = sum(product_prices)
total_price
```

```python
#A website wants to count how many unique visitors it had in a day.

visits = [101, 102, 103, 101, 104, 102, 105, 103, 105, 106]

#Find how many unique visitors visited the site.
unique_visitor = set(visits)
print(len(unique_visitor))
```

```python
#Count Distinct Words in a Sentence:

#A user enters a sentence (as a single string).
sentence = input("Enter a sentence:")

#Your task is to count how many distinct words are in the sentence
words = sentence.split()
distinct_words = set(words)
print(len(distinct_words))
```

```python
students = {
    "Nawal": [70, 80, 75],
    "Natasha": [85, 90, 88]
}

# Add three new students
students.update({
    "Ahmed": [60, 70, 65],
    "Ayesha": [90, 85, 95],
    "Hamza": [75, 80, 70]
})

# Calculate average marks
for name in students:
    average = sum(students[name]) / 3
    students[name].append(average)

print(students)
```

```python
# Get student information
name = input("Enter student name: ")
age = int(input("Enter student age: "))

# Store name and age in a tuple
personal_info = (name, age)

# Get subjects
subjects = input("Enter subjects separated by comma: ").split(",")

# Use set to keep subject names unique
subjects = set(subjects)

# Get marks for each subject
marks = []

for subject in subjects:
    mark = int(input("Enter marks for " + subject + ": "))
    marks.append(mark)

# Store information in a dictionary
student = {
    "Personal Info": personal_info,
    "Subjects": subjects,
    "Marks": marks
}

# Print summary report
print("\n--- Student Report ---")
print("Name:", personal_info[0])
print("Age:", personal_info[1])
print("Subjects:", subjects)
print("Marks:", marks)
print("Total Marks:", sum(marks))
print("Average Marks:", sum(marks) / len(marks))
```
