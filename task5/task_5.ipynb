```python
# 1. Write a function prime_upto(n) that returns all prime numbers up to n with input validation.
def prime_upto(n):
    if not isinstance(n, int):
        raise TypeError("Input must be an integer.")
    if n < 2:
        raise ValueError("Input must be at least 2.")
    
    primes = []
    for num in range(2, n + 1):
        is_prime = True
        for i in range(2, int(num**0.5) + 1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            primes.append(num)
    return primes
print("Primes up to 50:", prime_upto(50))
```

```python
# 2. Write a function that accepts a list of numbers and returns a dictionary of metrics.
def analyze_numbers(numbers):
    if not numbers:
        return {"total": 0, "average": 0, "max": None, "min": None, "all even numbers": []}
    
    return {
        "total": sum(numbers),
        "average": sum(numbers) / len(numbers),
        "max": max(numbers),
        "min": min(numbers),
        "all even numbers": [num for num in numbers if num % 2 == 0]
    }
print("Number Analysis:", analyze_numbers([10, 15, 22, 33, 40]))
```

```python
#Email Filtering System
emails = ["john@company.com", "alice@gmail.com", "bob@company.com", "john@company.com"]
target_domain = "@company.com"

# Extract all emails from a specific domain using list comprehension
company_emails = [email for email in emails if email.endswith(target_domain)]
print("Company Emails:", company_emails)

# Count how many unique users (before the @) there are using set comprehension
unique_users = {email.split('@')[0] for email in company_emails}
print("Unique Users Count:", len(unique_users))
print("Unique Users Set:", unique_users)
```
