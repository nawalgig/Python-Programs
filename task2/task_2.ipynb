```python
#Gym Membership Fee Calculator

registration_fee = int(input("Enter registration fee: "))
monthly_fee = int(input("Enter monthly fee: "))
months = int(input("Enter number of months you want to pay for: "))

total_cost = registration_fee + (monthly_fee * months)

has_coupon = input("Do you have a discount coupon? (yes/no): ")

discount_multiplier = 1.0 - (int(has_coupon == "yes") * 0.1)

total_cost *= discount_multiplier

print(f"Final amount to be paid: {total_cost}")
```

```python
#Simple E-Wallet Recharge and Spending:

wallet_balance = 0

money_added = float(input("How much money do you want to add? "))
food_expense = float(input("How much money did you spend on food? "))
transport_expense = float(input("How much money did you spend on transport? "))
shopping_expense = float(input("How much money did you spend on shopping? "))

wallet_balance += money_added       # Adding money
wallet_balance -= food_expense      # Deducting food expense
wallet_balance -= transport_expense # Deducting transport expense
wallet_balance -= shopping_expense  # Deducting shopping expense

#Calculate if the balance is between 500 and 5000 using logical operators
is_within_range = (wallet_balance > 500) and (wallet_balance < 5000)

print(is_within_range)
print(f"Final wallet balance: {wallet_balance}")
```

```python
#Expiry Date Tracker

from datetime import datetime, timedelta

# Ask the user for manufacture date and validity days
mfg_input = input("Enter manufacture date (YYYY-MM-DD): ")
validity_days = int(input("Enter validity period in days: "))

# Convert the text input into a real Python datetime object
mfg_date = datetime.strptime(mfg_input, "%Y-%m-%d")

# Calculate expiry date by adding validity days to the manufacture date
expiry_date = mfg_date + timedelta(days=validity_days)

# Get today's current date
today = datetime.now()

# Check whether the product is expired or valid today
is_expired = today > expiry_date

# Display results formatted cleanly using strftime()
# %B -> Full month name, %d -> Day number, %Y -> 4-digit Year
print(f"Manufacture Date : {mfg_date.strftime('%B %d, %Y')}")
print(f"Expiry Date      : {expiry_date.strftime('%B %d, %Y')}")

if is_expired:
    print("Status           : EXPIRED (Do not use!)")
else:
    print("Status           : VALID (Safe to use)")
```

```python
#Code Performance Timer

import time

n = int(input("Enter a number n: "))

start_time = time.time()

sum_of_squares = (n * (n + 1) * (2 * n + 1)) // 6

end_time = time.time()

execution_time_seconds = end_time - start_time

# Convert seconds to milliseconds
execution_time_milliseconds = execution_time_seconds * 1000

print(f"Computed Sum of Squares : {sum_of_squares}")
print(f"Execution Time (Sec)    : {execution_time_seconds:.9f} seconds")
print(f"Execution Time (Ms)     : {execution_time_milliseconds:.4f} ms")
```

```python
#OTP Verification System

import random
import time

amount = input("Enter the Amount to send (Rs.): ")

# Generate a 6-digit random OTP
otp = random.randint(100000, 999999)

print(f"SIMULATED SMS: Your OTP for sending Rs. {amount} is: {otp}")
print("Note: This OTP is valid for 10 seconds only!")

# Start the expiration clock
start_time = time.time()

user_otp = input("Enter the 6-digit OTP: ")

# Stop the clock
end_time = time.time()

time_taken = end_time - start_time

# Check 1: Did the user type the correct OTP?
# Check 2: Was it completed in under 10 seconds?
if user_otp.isdigit() and int(user_otp) == otp and time_taken <= 10:
    print(f"\n Amount Sent! Rs. {amount} has been transferred successfully.")
    print(f"(Verification took {time_taken:.2f} seconds)")
else:
    print("\n OTP Expired or Incorrect!")
    if time_taken > 10:
        print(f"Reason: You took too long! ({time_taken:.2f} seconds elapsed. Limit was 10s).")
    else:
        print("Reason: The OTP code you typed did not match.")
```
