```python
#We have four variables a = 5, b = 10, c = 3 and d = 6 swap the values of variable a & b and c & d.

a = 5
b = 10
c = 3
d = 6

a, b = b, a
c, d = d, c

print(f"a: {a}, b: {b}, c: {c}, d: {d}")
```

```python
#Ask the user for their name, age, and city and print the sentence in reverse order using .format()

name = input("Enter your name: ")  # e.g., Ali
age = input("Enter your age: ")    # e.g., 22
city = input("Enter your city: ")  # e.g., Lahore

print("{2} is where I live. {1} is my age. My name is {0}.".format(name, age, city))
```

```python
#Ask the user to input product and its price then generate a receipt using f-string:

product = input("Enter product name: ")
price = float(input("Enter product price: "))

print("\n" + "=" * 25)
print(f"{'RECEIPT':^25}") 
print("-" * 25)
print(f"Item: {product}")
print(f"Price: Rs. {price:.2f}") 
print("=" * 25)
```
