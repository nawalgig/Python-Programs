```python
#Use a for loop to print all numbers divisible by 7 from 1 to 100:
for i in range(1, 101):
    if i % 7 == 0:
        print(i)
```

```python
#Calculate and print the sum of even numbers using a for loop from 1 to 50:
total_sum = 0
for i in range(1, 51):
    if i % 2 == 0:
        total_sum += i
print("Sum of even numbers:", total_sum)
```

```python
#Ask the user for a password in a while loop. Allow 3 tries, and if the correct password is entered, 
#print "Access granted", otherwise print "Blocked":
correct_password = "secretpassword"
tries = 3

while tries > 0:
    password = input("Enter password: ")
    if password == correct_password:
        print("Access granted")
        break
    else:
        tries -= 1
        if tries == 0:
            print("Blocked")
```

```python
#Create a countdown from 10 to 1 using a while loop. 
# Use break if it reaches 5 and print "Midpoint reached":
n = 10
while n >= 1:
    if n == 5:
        print("Midpoint reached")
        break
    print(n)
    n -= 1
```

```python
#From 1 to 500, use a loop to find the first number divisible by both 13 and 7. Use break:
for i in range(1, 501):
    if i % 13 == 0 and i % 7 == 0:
        print("First number divisible by both 13 and 7:", i)
        break
```

```python
#From a list of words, create a dictionary where key is the word and value is the length,
# but only for words longer than 4 letters:
words = ["apple", "cat", "banana", "dog", "elephant"]
word_lengths = {word: len(word) for word in words if len(word) > 4}
print(word_lengths)
```

```python
#Given a string, build a dictionary using comprehension 
# where the keys are characters and values are their frequency in the string:
s = "programming"
char_freq = {char: s.count(char) for char in s}
print(char_freq)
```

```python
#Create a list from 100 to 150 using list comprehension 
# where prime numbers stay as-is, and non-primes become "Not Prime":
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

result_list = [n if is_prime(n) else "Not Prime" for n in range(100, 151)]
print(result_list)
```

```python
#Student Grade System
scores_list = []
total_students = 0

while True:
    name_input = input("Enter student name (or type 'done' to finish): ")
    if name_input.lower() == "done":
        break
    
    score_input = float(input("Enter score: "))
    if score_input < 0 or score_input > 100:
        continue
    
    scores_list.append(score_input)
    total_students += 1

if total_students > 0:
    avg_score = sum(scores_list) / total_students
    highest_score = max(scores_list)
    lowest_score = min(scores_list)
    
    print(f"Total number of students: {total_students}")
    print(f"Average score: {avg_score:.2f}")
    print(f"Highest score: {highest_score}")
    print(f"Lowest score: {lowest_score}")
else:
    print("No valid student scores entered.")
```

```python
#List Categorizer
l = list(range(1, 21))

even_list = []
odd_list = []
multiples_of_5 = []

for num in l:
    if num % 2 == 0:
        even_list.append(num)
    else:
        odd_list.append(num)
        
    if num % 5 == 0:
        multiples_of_5.append(num)

print("Even:", even_list)
print("Odd:", odd_list)
print("Multiples of 5:", multiples_of_5)
```

```python
#FizzBuzz Dictionary
fizz_buzz_dict = {}

for i in range(1, 31):
    if i % 3 == 0 and i % 5 == 0:
        fizz_buzz_dict[i] = "FizzBuzz"
    elif i % 3 == 0:
        fizz_buzz_dict[i] = "Fizz"
    elif i % 5 == 0:
        fizz_buzz_dict[i] = "Buzz"
    else:
        fizz_buzz_dict[i] = "None"

print(fizz_buzz_dict)
```
